```python
query = """
SELECT
    SUM(od.SalesAmount) AS TotalRevenue,
    SUM(od.Profit) AS TotalProfit
FROM Orders o
JOIN OrderDetails od
    ON o.OrderID = od.OrderID
WHERE o.OrderStatus = 'Delivered';
"""
```

```python
# Create Analytical Dataset

df = orders.merge(
    customers,
    on="CustomerID",
    how="left"
)
```

```python
# Add OrderDetails

df = df.merge(
    order_details,
    on="OrderID",
    how="inner"
)
```

```python
# Add Product Information

df = df.merge(
    products,
    on="ProductID",
    how="left"
)
```

```python
df.shape
```

```python
df.head()
```

```python
df.columns
```

```python
# Check Duplicate Order Details

df["OrderDetailID"].duplicated().sum()
```

```python
# Create Useful Date Columns

df["Year"] = df["OrderDate"].dt.year

df["Month"] = df["OrderDate"].dt.month

df["MonthName"] = df["OrderDate"].dt.month_name()

df["YearMonth"] = df["OrderDate"].dt.to_period("M").astype(str)
```

```python
# Profit Margin

df["ProfitMarginPct"] = (
    df["Profit"] /
    df["SalesAmount"]
) * 100
```
